# Plot learned KAN edge functions

Each KAN edge is a learned 1-D function. This notebook trains a two-input
model on `sin(pi x0) + 0.6 x1^2` and plots those functions with `kanx.viz`.

`plot_edges` writes a static figure (needs matplotlib). `write_edge_viewer`
writes `edges.html`, a layer picker with a hover readout, and does not need
matplotlib.


```python
import numpy as np

from kanx import KAN, fit_grid_to_data, set_global_seed
from kanx.viz import edge_curves, plot_edges, write_edge_viewer

set_global_seed(0)
rng = np.random.default_rng(0)
X = rng.uniform(-1, 1, size=(1024, 2)).astype('float32')
y = (np.sin(np.pi * X[:, :1]) + 0.6 * X[:, 1:2] ** 2).astype('float32')

model = KAN([2, 1])
model(np.zeros((1, 2), dtype='float32'))
fit_grid_to_data(model, X)
model.fit(X, y, epochs=30, batch_size=64, verbose=0)

x, curves = edge_curves(model.layers[0], n=200)
print(x.shape, curves.shape)  # (200,), (in_features, out_features, 200)
```

```python
fig = plot_edges(model, path='edges.png')
fig
write_edge_viewer(model, 'edges.html')
print('wrote edges.png and edges.html')
```
